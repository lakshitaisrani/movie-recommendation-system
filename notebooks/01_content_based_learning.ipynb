```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
ps.stem("playing")
```

```python
words = ["playing", "played", "plays", "loving", "studies"]

for word in words:
    print(word, "---->", ps.stem(word))
```

```python
def stem_text(text):
    words = text.split()
    stemmed_words = []

    for word in words:
        stemmed_words.append(ps.stem(word))

    return " ".join(stemmed_words)
```

```python
stem_text("playing played plays loving")
```

```python
movies_merged["tags"] = movies_merged["tags"].apply(stem_text)
```

```python
movies_merged[["title","tags"]].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000 , stop_words="english")
```

```python
vectors = cv.fit_transform(movies_merged["tags"]).toarray()
```

```python
vectors.shape
```

```python
cv.get_feature_names_out()
```

```python
vectors[0]
```

```python
movies_merged[movies_merged["tags"].str.contains("000",regex= False)][["title","tags"]].head()
```

```python
#removing numbers from column of vectors (5000 features)
import re
def remove_numbers(text):
    return re.sub(r'\b\d+\b', '', text)

movies_merged["tags"] = movies_merged["tags"].apply(remove_numbers)
```

```python
vectors = cv.fit_transform(movies_merged["tags"]).toarray()
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
similarity[0]
```

```python
movies_merged.iloc[0]["title"]
```

```python
sorted(list(enumerate(similarity[0])),reverse = True, key=lambda x: x[1])
#enumerate() attaches each movie's index to its similarity score.
#sorted() sorts the pairs.
#reverse=True puts the highest scores first.
#key=lambda x: x[1] sorts using the similarity score, not the movie index.
```

```python
movie_list = sorted(
    list(enumerate(similarity[0])),
    reverse=True,
    key=lambda x: x[1]
)
```

```python
for i in movie_list[1:11]:
    print(movies_merged.iloc[i[0]]["title"],i[1])
```

```python
def recommend(movie):
    if movie not in movies_merged["title"].values:
        print("Movie not found in our dataset!")
        return
    movie_index = movies_merged[movies_merged["title"] == movie].index[0]
    distances = similarity[movie_index]
    movie_list = sorted(
    list(enumerate(distances)),
    reverse=True,
    key=lambda x: x[1]
    )[1:11]
    for i in movie_list:
        print(movies_merged.iloc[i[0]]["title"])

recommend('Avatar')
```

```python
def recommend_saved(movie):
    if movie not in movies["title"].values:
        print("Movie not found in our dataset!")
        return

    movie_index = movies[movies["title"] == movie].index[0]
    distances = similarity_matrix[movie_index]

    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:11]

    for i in movie_list:
        print(movies.iloc[i[0]]["title"])
```
